#Funnel Analysis

```python
#Funnel müştərinin alış prosesində keçdiyi mərhələləri göstərir və harada itki olduğunu tapmaq üçün istifadə olunur.
```

```python
# Superstore dataset-i event-level məlumatlar (visit, add-to-cart, checkout)
# saxlamadığı üçün funnel mərhələlərini mövcud sütunlardan proxy olaraq yaratmışam.
# Bu yanaşma real user davranışını deyil, biznes məntiqinə əsaslanan təxmini funnel analizini göstərir.
```

```python
purchase = df[df['Sales'] > 0]['Customer ID'].unique() #satis
checkout = df[df['Profit'] > 0]['Customer ID'].unique() #kassa
cart = df[df['Discount'] > 0]['Customer ID'].unique()  #kart

funnel = {
    'Visited Product Page': len(df['Customer ID'].unique()), #her musteri azi birdefe mehsula baxib
    'Added to Cart': len(set(cart)),
    'Checkout Started': len(set(cart) & set(checkout)),
    'Purchase Completed': len(set(cart) & set(checkout) & set(purchase))
}


# DataFrame şəklində
funnel_df = pd.DataFrame.from_dict(funnel, orient='index', columns=['Users'])

# Drop-off faizi
#Drop-off = 1 − (Next step users / Current step users)
funnel_df['Drop-off %'] = 1 - funnel_df['Users'].shift(-1) / funnel_df['Users']
funnel_df['Drop-off %'] = funnel_df['Drop-off %'].fillna(0) * 100

print(funnel_df)
```

```python
# Əgər Added to Cart → Checkout drop-off böyükdürsə:
# Qiymət problemi
# UX problemi
# Əgər Checkout → Purchase drop-off böyükdürsə:
# Ödəniş
# Etibar problemi
```

#RFM

```python
# RFM nədir? müştəriləri davranışına görə segmentləşdirmək üçün istifadə olunan metoddur.

# R (Recency) – müştəri nə qədər yaxın zamanda alıb?

# F (Frequency) – neçə dəfə alıb?

# M (Monetary) – nə qədər pul xərcləyib?
#Tez-tez, yaxın zamanda və çox pul xərcləyən müştəri = ən dəyərli müştəri
```

```python
snapshot_date = df['Order Date'].max() + pd.Timedelta(days=1)

rfm = df.groupby('Customer ID').agg({
    'Order Date': lambda x: (snapshot_date - x.max()).days,  #Recency
    'Order ID': 'nunique',  #Frequency
    'Sales': 'sum'  #Monetary
}).reset_index()

rfm.columns = ['Customer ID', 'Recency', 'Frequency', 'Monetary']
```

```python
rfm
```

```python
rfm['R_Score'] = pd.qcut(rfm['Recency'], 5, labels=[5,4,3,2,1])
rfm['F_Score'] = pd.qcut(rfm['Frequency'].rank(method='first'), 5, labels=[1,2,3,4,5])
rfm['M_Score'] = pd.qcut(rfm['Monetary'], 5, labels=[1,2,3,4,5])
```

```python
rfm
```

```python
rfm['RFM_Segment'] = (
    rfm['R_Score'].astype(str) +
    rfm['F_Score'].astype(str) +
    rfm['M_Score'].astype(str)
)
rfm
```

```python
rfm[['R_Score','F_Score','M_Score']] = rfm[['R_Score','F_Score','M_Score']].astype(int)
```

```python
# Champions → ən dəyərli müştərilər

# Loyal → aktiv və stabil

# At Risk → gedə bilər

# Others → normal
```

```python
def segment(row):
    if row['R_Score'] == 5 and row['F_Score'] >= 4:
        return 'Champions'
    elif row['R_Score'] >= 4:
        return 'Loyal'
    elif row['R_Score'] <= 2 and row['F_Score'] >= 3:
        return 'At Risk'
    else:
        return 'Others'

rfm['Segment'] = rfm.apply(segment, axis=1)
```

#CLV

```python
#CLV (Customer Lifetime Value) → bir müştərinin şirkət üçün gələcəkdə gətirəcəyi potensial ümumi gəlir
```

```python
# Sadə CLV formula: AOV * Frequency * Retention
rfm_churn['AOV'] = rfm_churn['Monetary'] / rfm_churn['Frequency']

# Retention: churn = True → 0, False → 1
rfm_churn['Retention'] = (~rfm_churn['churned']).astype(int)

rfm_churn['CLV'] = rfm_churn['AOV'] * rfm_churn['Frequency'] * rfm_churn['Retention']

# Segment üzrə ortalama CLV
segment_clv = rfm_churn.groupby('Segment')['CLV'].mean().sort_values(ascending=False)
segment_clv
```

#A/B Test

```python
#Others müştəriləri xilas etmək
```

```python
import numpy as np

others = rfm_churn[rfm_churn['Segment'] == 'Others'].copy()

np.random.seed(42)
others['group'] = np.random.choice(['A', 'B'], size=len(others))
```

```python
np.random.seed(42)

others['converted'] = np.where(
    others['group'] == 'B',
    np.random.binomial(1, 0.30, len(others)),  # B qrupunda 30% conversion ,Təklif alanlar daha çox alış edir (hipotez)
    np.random.binomial(1, 0.20, len(others))   # A qrupunda 20% conversion
)
```

```python
conversion_rates = others.groupby('group')['converted'].mean()
conversion_rates
```

```python
from statsmodels.stats.proportion import proportions_ztest

success = others.groupby('group')['converted'].sum()
nobs = others.groupby('group')['converted'].count()

z_stat, p_value = proportions_ztest(success, nobs)
p_value  #0.05<
```
